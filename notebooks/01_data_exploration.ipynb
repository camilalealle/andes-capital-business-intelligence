```python
import pandas as pd
from pathlib import Path

# Define project directories
project_dir = Path.cwd()

if not (project_dir / "data").exists():
    project_dir = project_dir.parent

raw_dir = project_dir / "data" / "raw"

# Load original datasets
sales = pd.read_csv(raw_dir / "hecho_ventas_propiedades.csv")
customers = pd.read_csv(raw_dir / "dim_clientes.csv")
properties = pd.read_csv(raw_dir / "dim_propiedades.csv")

print("Datasets loaded successfully!")
```

```python
# Initial overview of the datasets

datasets = {
    "Sales": sales,
    "Customers": customers,
    "Properties": properties
}

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
# Preview the first records of each dataset

for name, df in datasets.items():
    print(f"\n{name.upper()}")
    display(df.head())
```

```python
# Inspect data types and non-null values

sales.info()
```

```python
# Inspect customers dataset

customers.info()
```

```python
# Inspect properties dataset

properties.info()
```

```python
# Check missing values

for name, df in datasets.items():
    print(f"\n{name.upper()}")
    print(df.isna().sum())
```

```python
# Inspect customer categories

customers["segmento_comprador"].value_counts(dropna=False)
```

```python
# Inspect categorical values for potential sentinels

categorical_columns = {
    "Sales": ["ciudad", "tipo_propiedad", "canal_venta"],
    "Customers": ["segmento_comprador", "pais", "ciudad"],
    "Properties": ["tipo_propiedad", "ciudad", "barrio", "categoria_propiedad"]
}

for name, columns in categorical_columns.items():
    print(f"\n{name.upper()}")

    for col in columns:
        print(f"\n{col}:")
        print(datasets[name][col].value_counts(dropna=False))
```

```python
# Inspect numerical variables

for name, df in datasets.items():
    print(f"\n{name.upper()}")
    display(df.describe())
```

```python
# Check common numeric sentinel values

sentinels = [-999, -99, -1, 9999, 99999, 999999]

for name, df in datasets.items():
    numeric = df.select_dtypes(include="number")

    print(f"\n{name.upper()}")
    print(numeric.isin(sentinels).sum())
```

```python
# Check duplicate records

for name, df in datasets.items():
    print(f"{name}: {df.duplicated().sum()} duplicate rows")
```

```python
# Check duplicate primary keys

primary_keys = {
    "Sales": "id_venta",
    "Customers": "id_cliente",
    "Properties": "id_propiedad"
}

for name, key in primary_keys.items():
    duplicates = datasets[name][key].duplicated().sum()
    print(f"{name}: {duplicates} duplicate IDs")
```

```python
# Check referential integrity

missing_customers = ~sales["id_cliente"].isin(customers["id_cliente"])

missing_properties = ~sales["id_propiedad"].isin(properties["id_propiedad"])

print(f"Sales with missing customers: {missing_customers.sum()}")
print(f"Sales with missing properties: {missing_properties.sum()}")
```

```python
# Check consistency between sales and properties

sales_properties = sales.merge(
    properties[["id_propiedad", "ciudad", "tipo_propiedad"]],
    on="id_propiedad",
    how="left",
    suffixes=("_sales", "_properties"),
    validate="many_to_one"
)

city_mismatches = (
    sales_properties["ciudad_sales"]
    != sales_properties["ciudad_properties"]
)

type_mismatches = (
    sales_properties["tipo_propiedad_sales"]
    != sales_properties["tipo_propiedad_properties"]
)

print(f"City mismatches: {city_mismatches.sum()}")
print(f"Property type mismatches: {type_mismatches.sum()}")
```

```python
# Validate commission calculations

expected_commission = (
    sales["precio_venta"] * sales["porcentaje_comision"]
)

commission_difference = (
    sales["monto_comision"] - expected_commission
).abs()

print(f"Maximum difference: {commission_difference.max():.2f}")
print(f"Records with differences greater than 1: {(commission_difference > 1).sum()}")
```

```python
# Investigate commission differences

commission_check = sales[
    ["id_venta", "precio_venta", "porcentaje_comision", "monto_comision"]
].copy()

commission_check["comision_calculada"] = expected_commission

commission_check["diferencia"] = commission_difference

display(
    commission_check.sort_values(
        "diferencia", ascending=False
    ).head(10)
)
```

```python
# Validate differences caused by commission rate rounding

rounding_tolerance = sales["precio_venta"] * 0.00005 + 0.5

outside_tolerance = commission_difference > rounding_tolerance

print(f"Records outside rounding tolerance: {outside_tolerance.sum()}")
print(f"Percentage within tolerance: {(~outside_tolerance).mean():.2%}")
```

```python
# Investigate records outside rounding tolerance

commission_issues = commission_check.loc[outside_tolerance].copy()

commission_issues["tolerancia"] = rounding_tolerance[outside_tolerance]

commission_issues["exceso"] = (
    commission_issues["diferencia"] - commission_issues["tolerancia"]
)

display(
    commission_issues.sort_values(
        "exceso", ascending=False
    ).head(15)
)
```

### Commission Data Quality Assessment

Commission amounts were validated against sale prices and recorded commission rates.

- Total transactions: 8,500
- Transactions within the assumed rounding tolerance: 8,469 (99.64%)
- Transactions exceeding the tolerance: 31 (0.36%)
- Maximum observed excess above tolerance: 0.38

The differences are consistent with a possible loss of decimal precision in recorded commission rates. However, the original calculation methodology is not documented.

**Decision:** Preserve the original commission amounts and document the rounding uncertainty. No records were removed or modified.

```python
# Validate sales dates

sales_dates = pd.to_datetime(
    sales["fecha_venta"],
    format="%Y-%m-%d",
    errors="coerce"
)

print(f"Invalid dates: {sales_dates.isna().sum()}")
print(f"First sale: {sales_dates.min()}")
print(f"Last sale: {sales_dates.max()}")
```

```python
# Check monthly transaction distribution

monthly_sales = (
    sales_dates
    .dt.to_period("M")
    .value_counts()
    .sort_index()
)

display(monthly_sales)
```

## Data Quality Assessment — Summary

The initial assessment covered three datasets containing 8,500 sales transactions, 3,500 customer records, and 8,000 property records.

### Key Findings

- No missing values were identified by Pandas.
- No full-row duplicates or duplicate primary keys were detected.
- No orphan customer or property references were found.
- Property locations and types are consistent across the sales and property tables.
- All sales dates are valid and fall between January 1, 2023, and December 31, 2024.
- All 24 months contain recorded transactions.
- Commission differences were identified, with 99.64% of records falling within the assumed rounding tolerance.

### Data Quality Decisions

- Preserve the original datasets without modification.
- Convert sales dates to datetime during data preparation.
- Retain recorded commission amounts and document the observed precision differences.
- Perform additional business-rule validations before finalizing the analytical model.

### Conclusion

The datasets show good initial structural integrity and are suitable for proceeding to the data preparation phase, subject to the remaining business-rule checks.
